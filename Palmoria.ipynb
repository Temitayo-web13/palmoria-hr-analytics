```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python

emp = pd.read_csv("Palmoria Group emp-data.csv")

emp.head()
```

```python
emp.shape
```

```python
emp.isnull().sum()
```

```python
for column in emp.columns:
    print("\n", column)
    print(emp[column].unique())
```

```python
emp.duplicated().sum()
```

```python
emp[emp.duplicated(keep=False)]
```

```python
emp = emp.drop_duplicates()
```

```python
emp.duplicated().sum()
```

```python
emp = emp[emp["Salary"].notna()]
```

```python
emp["Salary"].isnull().sum()
```

```python
emp = emp[emp["Department"].notna()]
```

```python
emp["Department"].isnull().sum()
```

```python
emp["Gender"] = emp["Gender"].fillna("Undisclosed")
```

```python
emp["Gender"].value_counts()
```

```python
text_columns = [
    "Name",
    "Gender",
    "Department",
    "Location",
    "Rating"
]

for col in text_columns:
    emp[col] = emp[col].astype(str).str.strip()
```

```python
print("Rows:", emp.shape[0])
print("Columns:", emp.shape[1])
```

```python
def salary_band(salary):
    lower = int(np.floor(salary / 10000) * 10000)
    upper = lower + 10000
    return f"${lower:,} - ${upper:,}"
```

```python
emp["Salary Band"] = emp["Salary"].apply(salary_band)
```

```python
emp[["Salary", "Salary Band"]].head(20)
```

```python
emp.head()
```

```python
emp["Meets $90k Requirement"] = np.where(
    emp["Salary"] >= 90000,
    "Yes",
    "No"
)
```

```python
emp["Meets $90k Requirement"].value_counts()
```

```python
emp.head()
```

```python
emp["Meets $90k Requirement"].value_counts(normalize=True) * 100
```

```python
bonus = pd.read_excel("Palmoria Group Bonus Rules.xlsx")

bonus.head(10)
```

```python
rating_columns = [
    "Very Poor",
    "Poor",
    "Average",
    "Good",
    "Very Good"
]

for col in rating_columns:
    bonus[col] = pd.to_numeric(
        bonus[col],
        errors="coerce"
    )
```

```python
bonus
```

```python
bonus_long = bonus.melt(
    id_vars="Department",
    var_name="Rating",
    value_name="Bonus %"
)
```

```python
bonus_long
```

```python
emp = emp.merge(
    bonus_long,
    on=["Department", "Rating"],
    how="left"
)
```

```python
emp.head()
```

```python
emp["Bonus %"].isnull().sum()
```

```python
emp["Bonus"] = emp["Salary"] * emp["Bonus %"]
```

```python
emp.head()
```

```python
emp["Bonus %"] = emp["Bonus %"].fillna(0)
```

```python
emp["Bonus"] = emp["Salary"] * emp["Bonus %"]
```

```python
emp["Total Compensation"] = (
    emp["Salary"] + emp["Bonus"]
)
```

```python
emp.head()
```

```python
emp.to_csv(
    "Palmoria_Cleaned_Employee_Data.csv",
    index=False
)
```

```python
# actual HR analysis
```

```python
# Gender distribution
gender_distribution = (
    emp["Gender"]
    .value_counts()
    .reset_index()
)

gender_distribution.columns = [
    "Gender",
    "Employees"
]

gender_distribution
```

```python
# Gender distribution %
gender_distribution["Percentage"] = (
    gender_distribution["Employees"]
    / gender_distribution["Employees"].sum()
    * 100
)

gender_distribution
```

```python
# Gender by region
gender_region = pd.crosstab(
    emp["Location"],
    emp["Gender"]
)

gender_region
```

```python
# Gender by region %
gender_region_pct = pd.crosstab(
    emp["Location"],
    emp["Gender"],
    normalize="index"
) * 100

gender_region_pct
```

```python
# Gender by department
gender_department = pd.crosstab(
    emp["Department"],
    emp["Gender"]
)

gender_department
```

```python
# Gender by department
gender_department_pct = pd.crosstab(
    emp["Department"],
    emp["Gender"],
    normalize="index"
) * 100

gender_department_pct
```

```python
# Performance ratings by gender
rating_gender = pd.crosstab(
    emp["Rating"],
    emp["Gender"]
)

rating_gender
```

```python
# Performance ratings by gender %
rating_gender_pct = pd.crosstab(
    emp["Rating"],
    emp["Gender"],
    normalize="columns"
) * 100

rating_gender_pct
```

```python
# Salary by gender
salary_gender = (
    emp.groupby("Gender")["Salary"]
    .agg(["count", "mean", "median", "min", "max"])
    .round(2)
)

salary_gender
```

```python
# Calculate gender pay gap
avg_male_salary = emp.loc[
    emp["Gender"] == "Male",
    "Salary"
].mean()

avg_female_salary = emp.loc[
    emp["Gender"] == "Female",
    "Salary"
].mean()

pay_gap = (
    (avg_male_salary - avg_female_salary)
    / avg_male_salary
) * 100

print(f"Male average salary: ${avg_male_salary:,.2f}")
print(f"Female average salary: ${avg_female_salary:,.2f}")
print(f"Gender pay gap: {pay_gap:.2f}%")
```

```python
# Pay gap by region
salary_region_gender = (
    emp.groupby(["Location", "Gender"])["Salary"]
    .mean()
    .reset_index()
)

salary_region_gender
```

```python
# Pay gap by region, To make it easier to compare:
salary_region_gender_pivot = (
    salary_region_gender
    .pivot(
        index="Location",
        columns="Gender",
        values="Salary"
    )
)

salary_region_gender_pivot
```

```python
# Pay gap by department
salary_department_gender = (
    emp.groupby(["Department", "Gender"])["Salary"]
    .mean()
    .reset_index()
)

salary_department_gender
```

```python
# Pay gap by department, To make it easier to compare:
salary_department_gender_pivot = (
    salary_department_gender
    .pivot(
        index="Department",
        columns="Gender",
        values="Salary"
    )
)

salary_department_gender_pivot
```
